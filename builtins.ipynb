# abs()
Returns the absoulte value of a number. (Not done for complex numbers here)

```python
def my_abs(number):
    if number >= 0:
        return number
    else:
        return number * -1

my_abs(-3) == abs(-3)
```

# all()
Return True if all elements of the iterable are true (or if the iterable is empty)

```python
def my_all(iterable):
    for element in iterable:
        if not element:
            return False
        
    return True
```

**Explanation:** "if" checks whether the statement is truthy. Therefore if the element per se is falsy, as all the elements: False, 0, 0.0, "", [], {}, None", the "not" turns it into a true element. The if statement is therefore fulfilled, which finally ends the loop by returning False.

# any()

```python
def my_any(iterable):
    if len(iterable) == 0:
        return False
    for element in iterable:
        if element:
            return True
        
    return False
```

Similar logic to all(). If any element in the iterable "variable" is truthy, which is checked by the if statement: if element, the loop is ended with True, else it will end with False

# bin()

```python
def my_bin(number):
    # check how many powers of 2 are needed
    list_bin = []
    i=0
    while number > 2**i:
        i = i+1

    for x in range(i, -1, -1): # to show that we start with i and go just before -1, i.e. 0,  with steps of -1, i.e. going the loop backwards
        if number // 2**x == 1:
            number = number - 2**x
            list_bin.append("1")
        else:
            list_bin.append("0")
    return "".join(list_bin)

x = my_bin(5)
print(x)
```

# bool()
Only for the following values: False, 0, 0.0, "", [], {}, None

```python
def my_bool(value):
    check_list = [False, 0, 0.0, None]
    if value in check_list:
        return False
    elif isinstance(value, (str, list, tuple, dict, set)):
        if len(value) == 0:
            return False
        else:
            return True
    else:
        return True

my_bool("hello, world")
```

# divmod()
return a pair of numbers consisting of their quotient and remainder when using integer division

```python
def my_divmod(x,y):
    quotient = x // y
    remainder = x % y
    return quotient, remainder

divmod(10, 3) == my_divmod(10, 3)
```

# enumerate()

```python
for i, character in enumerate("asdfghjklöä"):
    print(i, character)
```

```python
def my_enumerate(iterable, start = 0):
    enumerate_tuple = []
    for i in range(len(iterable)):
        enumerate_tuple.append((i + start, iterable[i]))

    return enumerate_tuple



list_bsp = ["apple", "banana", "coconut"]
for i, character in my_enumerate(list_bsp, 4):
    print(i, character)


my_enumerate(list_bsp, 4) == list(enumerate(list_bsp, 4))
```

# filter()

```python
def my_filter(function, iterable):
    true_false_list = []
    for i, element in my_enumerate(iterable):
        fkt_value = function(element)
        true_false_list.append((i, fkt_value))

    new_iterable = []
    for tuple in true_false_list:
        if tuple[1] == True:
            new_iterable.append(iterable[tuple[0]])

    return new_iterable



numbers = [3, 8, 11, 14, 17, 20, 25, 30]

def bigger_than_15(x):
    return x > 15

my_filter(bigger_than_15, numbers) == list(filter(bigger_than_15, numbers))
```

# float()

```python
def my_float(string):
    new_float = 0
    real_string = str(string)

    is_negative = real_string.startswith("-")
        
    if real_string.startswith("-"):
        real_string = real_string.replace("-", "")

    if "." not in real_string:
        complete_string = f"{real_string}.0"
    else:
        complete_string = real_string

    point_index = complete_string.index(".")

    # for the numbers before the period
    decimal = point_index -1
    for i in range(0, point_index):
        new_float += int(real_string[i]) * 10**(decimal)
        decimal -= 1

    # for the numbers after the period

    decimal_after = -1
    rounding = 0
    for k in range(point_index + 1, len(real_string)):
        new_float += int(real_string[k]) * 10**(decimal_after)
        decimal_after -= 1
        rounding += 1
        new_float = round(new_float, rounding)

    
    if is_negative:
        new_float *= -1


    return new_float




my_float("-3.1415")

```

```python
def my_int(string):
    new_int = 0

    # create and clean the string
    clean_string = str(string).strip()

    if "." in clean_string:
        point_index = clean_string.find(".")
        clean_string = clean_string[:point_index]

    if clean_string.startswith("-"):
        is_negative = True
        clean_string = clean_string.replace("-", "")

    # print(clean_string)

    # get the length
    k = len(clean_string)

    # print(k)
    # 12345678
    
    power = k-1
    # decompose the number into 10^x steps
    for i in clean_string:
        # use 10^x to reassemble the structure
        new_int += (ord(i) - ord("0")) * 10**power # ord() gives the unicode number (which is a integer). The unicode numbers for 0-9 come one after another, meaning with -ord("0") the real integer is made
        power -= 1
        print(new_int)

    if is_negative:
        new_int *= -1

    return new_int



my_int(-20032.0)
```

# len()

```python
def my_len(iterable):
    base = 0
    try:
        for i in iterable:
            base += 1
    except TypeError:
        return TypeError
                
    return base

my_len(1234567)
```

# map()

```python
def my_map(function, iterable):
    mapped_items = []
    try:
        for i in iterable:
            mapped_items.append(function(i))

    except TypeError:
        return TypeError

    return mapped_items


print(my_map(str.upper, "hello"))
print(my_map(lambda x: x * 2, [1, 2, 3]))
print(my_map(str, (1, 2, 3)))
```

# pow()

```python
def my_pow(base: int, exp: int):
    
    result = 1
    if exp >= 0:
        for i in range(exp):
            result *= base

    else:
        for i in range(abs(exp)):
            result /= base

    return result

my_pow(13, 38)
```

# range()

```python
def my_range(start = 0, stop = None, step = 1):
    # print(start)
    # print(stop)
    # print(step)

    if step == 0:
        return ValueError
    if type(start) and type(stop) and type(step) != int:
        return TypeError
    
    if stop == None: # there was only one argument given in the ()
        stop = start
        start = 0


    range_list = []

    while (step > 0 and start < stop) or (step < 0 and start > stop):
        range_list.append(start)
        start += step

    return range_list

my_range(13.1, 0, 0)
```
